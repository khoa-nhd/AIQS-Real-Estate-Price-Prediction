```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/vietnam_housing_dataset.csv")

df.head()
```

```python
df["Price"].dtype 
```

```python
(df["Price"] <= 0).sum() 
```

```python
df["Price"].describe() 
```

```python
price = df["Price"].dropna()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(price, bins=50)

plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Distribution of House Prices")

plt.show()
```

```python
mean_price = price.mean()
median_price = price.median()

print(f"Mean Price: {mean_price}")
print(f"Median Price: {median_price}")
```

```python
plt.figure(figsize=(10, 6))

plt.hist(price, bins=50)

plt.axvline(
    mean_price,
    linestyle="--",
    label=f"Mean = {mean_price}"
)

plt.axvline(
    median_price,
    linestyle="--",
    label=f"Median = {median_price}"
)

plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Price Distribution with Mean and Median")
plt.legend()

plt.show()
```

```python
quantiles = price.quantile(
    [0.01, 0.05, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
)

quantiles
```

```python
skewness = price.skew()

print(f"Skewness of Price: {skewness}")
```

```python
plt.figure(figsize=(10, 3))

plt.boxplot(price, orientation="horizontal")

plt.xlabel("Price")
plt.title("Boxplot of House Prices")

plt.show()
```
